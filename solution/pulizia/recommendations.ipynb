# Cleaning recommendations.csv

```python
import pandas as pd
import os
```

```python
raw_path = os.path.join("../..", "dataset", "raw", "recommendations.csv")
out_path = os.path.join("../..", "dataset", "cleaned", "recommendations_cleaned.csv")
```

```python
# Load dataset
recs = pd.read_csv(raw_path)
original_rows = len(recs)

# Keep only real columns
recs = recs[['mal_id', 'recommendation_mal_id']]

# Drop rows without valid IDs
recs = recs.dropna(subset=['mal_id', 'recommendation_mal_id'])

# Convert IDs to numeric
recs[['mal_id', 'recommendation_mal_id']] = (
    recs[['mal_id', 'recommendation_mal_id']]
    .apply(pd.to_numeric, errors='coerce')
    .astype(int)
)

print("Rows removed:", original_rows - len(recs))

# Save cleaned dataset
recs.to_csv(out_path, index=False)
```
