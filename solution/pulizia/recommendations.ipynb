# Cleaning recommendations.csv

```python
import pandas as pd
import os
```

```python
RAW_PATH = os.path.join("../..", "dataset", "raw", "recommendations.csv")
CLEAN_PATH = os.path.join("../..", "dataset", "cleaned", "recommendations_cleaned.csv")
```

```python
df = pd.read_csv(RAW_PATH)
original_rows = len(df)

# Convertiamo entrambi gli ID in numerico
df['mal_id'] = pd.to_numeric(df['mal_id'], errors='coerce').astype(int)
df['recommendation_mal_id'] = pd.to_numeric(df['recommendation_mal_id'], errors='coerce').astype(int)
# Rimuoviamo solo righe senza uno dei due ID
df = df.dropna(subset=['mal_id', 'recommendation_mal_id'])
# Rimuoviamo solo righe identiche
df = df.drop_duplicates()

print("Righe iniziali:", original_rows)
print("Righe finali:", len(df))
print("Righe rimosse:", original_rows - len(df))

df.to_csv(CLEAN_PATH, index=False, encoding="utf-8")
```
