# Cleaning person_alternate_names

```python
import os
import pandas as pd
```

```python
RAW = os.path.join (".." , ".." , "dataset" , "raw" , "person_alternate_names.csv")
CLEAN = os.path.join(".." , ".." , "dataset" , "cleaned" , "person_alternate_names_cleaned.csv")
```

```python
#load
df = pd.read_csv(RAW)
original_rows = len(df)
df.head()
```

```python
#id must be numeric
df['person_mal_id'] = pd.to_numeric(df['person_mal_id'], errors='coerce')
df = df.dropna(subset=['person_mal_id'])
df['person_mal_id'] = df['person_mal_id'].astype(int)

#clean alt_name
df = df.dropna(subset=["alt_name"])
df['alt_name'] = df['alt_name'].astype(str).str.strip()
df = df[(df['alt_name'] != "")]

#remove duplicates
before = len(df)
df = df.drop_duplicates(subset=["person_mal_id" , "alt_name"])
print("removed duplicates:" , before - len(df))

#final check: missing values in key columns
print(df[["person_mal_id", "alt_name"]].isna().sum())

#save
df.to_csv(CLEAN, index=False)
print("rows before:" , original_rows, "| rows after:" , len(df))
```
