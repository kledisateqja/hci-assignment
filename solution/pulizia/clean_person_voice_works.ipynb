```python
import pandas as pd
import os
```

```python
# path
RAW = os.path.join (".." , ".." , "dataset" , "raw" , "person_voice_works.csv")
CLEAN = os.path.join(".." , ".." , "dataset" , "cleaned" , "person_voice_works_cleaned.csv")
```

```python
#load
df = pd.read_csv(RAW)
original_rows = len(df)
df.head()
```

```python
# person id numeric
df["person_mal_id"] = pd.to_numeric(df["person_mal_id"], errors="coerce")
df = df.dropna(subset=["person_mal_id"])
df["person_mal_id"] = df["person_mal_id"].astype(int)

# anime id numeric
df["anime_mal_id"] = pd.to_numeric(df["anime_mal_id"], errors="coerce")
df = df.dropna(subset=["anime_mal_id"])
df["anime_mal_id"] = df["anime_mal_id"].astype(int)

# character id numeric
df["character_mal_id"] = pd.to_numeric(df["character_mal_id"], errors="coerce")
df = df.dropna(subset=["character_mal_id"])
df["character_mal_id"] = df["character_mal_id"].astype(int)

# clean role
df["role"] = df["role"].astype(str).str.strip()
df.loc[df["role"].str.lower().isin(["", "nan"]), "role"] = pd.NA

# clean language
df["language"] = df["language"].astype(str).str.strip()
df.loc[df["language"].str.lower().isin(["", "nan"]), "language"] = pd.NA

#remove duplicates
before = len(df)
df = df.drop_duplicates(subset=["person_mal_id", "anime_mal_id", "character_mal_id", "role", "language"])
print("removed duplicates:", before - len(df))

#save
df.to_csv(CLEAN, index=False)
print("rows before:", original_rows, "| rows after:", len(df))
```
