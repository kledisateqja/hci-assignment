```python
import pandas as pd
import os
```

```python
# path
RAW = os.path.join (".." , ".." , "dataset" , "raw" , "characters.csv")
CLEAN = os.path.join(".." , ".." , "dataset" , "cleaned" , "characters_cleaned.csv")
```

```python
#load
df = pd.read_csv(RAW)
original_rows = len(df)
df.head()
```

```python
#id must be numeric
df["character_mal_id"] = pd.to_numeric(df["character_mal_id"], errors="coerce")
df = df.dropna(subset=["character_mal_id"])
df["character_mal_id"] = df["character_mal_id"].astype(int)

#clean name
df = df.dropna(subset=["name"])
df["name"] = df["name"].astype(str).str.strip()
df = df[(df["name"] != "")]
df = df[(df["name"].str.lower() != "nan")]

#clean url
df["url"] = df["url"].astype(str).str.strip()
df.loc[df["url"].str.lower().isin(["", "nan"]), "url"] = pd.NA

#clean name_kanji
df["name_kanji"] = df["name_kanji"].astype(str).str.strip()
df.loc[df["name_kanji"].str.lower().isin(["", "nan"]), "name_kanji"] = pd.NA

#clean image
df["image"] = df["image"].astype(str).str.strip()
df.loc[df["image"].str.lower().isin(["", "nan"]), "image"] = pd.NA

#favorites numeric
df["favorites"] = pd.to_numeric(df["favorites"], errors="coerce")

#clean about
df["about"] = df["about"].astype(str).str.strip()
df.loc[df["about"].str.lower().isin(["", "nan"]), "about"] = pd.NA

#remove duplicates
before = len(df)
df = df.drop_duplicates(subset=["character_mal_id"])
print("removed duplicates:", before - len(df))

#save
df.to_csv(CLEAN, index=False)
print("rows before:", original_rows, "| rows after:", len(df))
```
