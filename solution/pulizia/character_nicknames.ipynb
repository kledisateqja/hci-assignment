# Cleaning of Character Nicknames

```python
import pandas as pd
import os
```

```python
# path
raw_path = os.path.join("../..", "dataset", "raw", "character_nicknames.csv")
out_path = os.path.join("../..", "dataset", "cleaned", "character_nicknames_cleaned.csv")
```

```python
# Load dataset
character_nicknames = pd.read_csv(raw_path)
original_rows = len(character_nicknames)

# remove rows without primary keys
character_nicknames = character_nicknames.dropna(subset=['character_mal_id'])
character_nicknames = character_nicknames.dropna(subset=['nickname'])
character_nicknames['character_mal_id'] = pd.to_numeric(character_nicknames['character_mal_id'], errors='coerce').astype(int)
character_nicknames = character_nicknames.drop_duplicates()

print("Rows removed:", original_rows - len(character_nicknames))

# Save cleaned dataset
character_nicknames.to_csv(out_path, index=False)
```
