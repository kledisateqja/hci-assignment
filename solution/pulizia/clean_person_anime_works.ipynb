Cleaning person_anime_works

```python
import os
import numpy as np
import pandas as pd
```

```python
#path
RAW = os.path.join("..", "..", "dataset", "raw" , "person_anime_works.csv")
CLEAN = os.path.join(".." , ".." , "dataset", "cleaned" , "person_anime_works_cleaned.csv")
```

```python
#load
df = pd.read_csv(RAW)
original_rows = len(df)
df.head()
```

```python
df.columns
df.count()
```

```python
# person id numeric
df["person_mal_id"] = pd.to_numeric(df["person_mal_id"], errors="coerce")
df = df.dropna(subset=["person_mal_id"])
df["person_mal_id"] = df["person_mal_id"].astype(int)

# anime id numeric
df["anime_mal_id"] = pd.to_numeric(df["anime_mal_id"], errors="coerce")
df = df.dropna(subset=["anime_mal_id"])
df["anime_mal_id"] = df["anime_mal_id"].astype(int)

# clean position
df["position"] = df["position"].astype(str).str.strip()
df.loc[df["position"].str.lower().isin(["", "nan"]), "position"] = pd.NA

#duplicates
before = len(df)
df = df.drop_duplicates(subset=["person_mal_id", "anime_mal_id", "position"])
print("removed duplicates:", before - len(df))

len(df)
df.count()

df.to_csv(CLEAN, index=False)
print("rows before:", original_rows, "| rows after:", len(df))

```
