# Cleaning person_details

```python
import os
import pandas as pd
import numpy as np
```

```python
RAW = os.path.join (".." , ".." , "data" , "raw" , "person_details.csv")
CLEAN = os.path.join(".." , ".." , "data" , "cleaned" , "person_details_cleaned.csv")
```

```python
df = pd.read_csv(RAW)
original_rows = len(df)
df.head()
```

```python
len(df)
df.count()
```

```python
# id must be numeric
df["person_mal_id"] = pd.to_numeric(df["person_mal_id"], errors = "coerce")
df = df.dropna(subset=["person_mal_id"])
df["person_mal_id"] = df["person_mal_id"].astype(int)


#columns
text_cols = ["url" , "website_url", "image_url", "name", "given_name", "family_name", "relevant_location"]

df["url"] = df["url"].astype(str).str.strip()
df.loc[df["url"].str.lower().isin(["", "nan"]), "url"] = pd.NA

df["website_url"] = df["website_url"].astype(str).str.strip()
df.loc[df["website_url"].str.lower().isin(["", "nan"]), "website_url"] = pd.NA

df["image_url"] = df["image_url"].astype(str).str.strip()
df.loc[df["image_url"].str.lower().isin(["", "nan"]), "image_url"] = pd.NA

df["name"] = df["name"].astype(str).str.strip()
df.loc[df["name"].str.lower().isin(["", "nan"]), "name"] = pd.NA

df["given_name"] = df["given_name"].astype(str).str.strip()
df.loc[df["given_name"].str.lower().isin(["", "nan"]), "given_name"] = pd.NA

df["family_name"] = df["family_name"].astype(str).str.strip()
df.loc[df["family_name"].str.lower().isin(["", "nan"]), "family_name"] = pd.NA

df["relevant_location"] = df["relevant_location"].astype(str).str.strip()
df.loc[df["relevant_location"].str.lower().isin(["", "nan"]), "relevant_location"] = pd.NA


#birthday
df["birthday"] = pd.to_datetime(df["birthday"] , errors = "coerce")

#favorites (numeric)
df["favorites"] = pd.to_numeric(df["favorites"], errors = "coerce").fillna(0).astype(int)

before = len(df)
df = df.drop_duplicates(subset=["person_mal_id"])
print("removed duplicates:", before - len(df))

#final check
len(df)
df.count()

#save
df.to_csv(CLEAN, index=False)
print("rows before:", original_rows, "| rows after:", len(df))
```
