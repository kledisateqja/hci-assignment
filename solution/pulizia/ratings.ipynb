# Cleaning ratings.csv

```python
import pandas as pd
import os
from unidecode import unidecode
```

```python
RAW_PATH = os.path.join("../..", "dataset", "raw", "ratings.csv")
CLEAN_PATH = os.path.join("../..", "dataset", "cleaned", "ratings_cleaned.csv")
```

```python
df = pd.read_csv(RAW_PATH)
original_rows = len(df)

df['username'] = df['username'].fillna("Unknown")
df['status'] = df['status'].fillna("Unknown")

def normalize_text(x):
    if isinstance(x, str):
        return unidecode(x).strip().lower()
    return x

df['username'] = df['username'].apply(normalize_text)
df['status'] = df['status'].apply(normalize_text)

df['anime_id'] = pd.to_numeric(df['anime_id'], errors='coerce').astype(int)
df = df.dropna(subset=['anime_id'])

# Convertiamo il voto in numerico
df['score'] = pd.to_numeric(df['score'], errors='coerce')
df = df.drop_duplicates()

print("Righe iniziali:", original_rows)
print("Righe finali:", len(df))
print("Righe rimosse:", original_rows - len(df))


df.to_csv(CLEAN_PATH, index=False, encoding="utf-8")
```
