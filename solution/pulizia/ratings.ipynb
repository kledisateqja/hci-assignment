# Cleaning ratings.csv

```python
import pandas as pd
import os
from unidecode import unidecode
```

```python
RAW_PATH = os.path.join("../..", "dataset", "raw", "ratings.csv")
CLEAN_PATH = os.path.join("../..", "dataset", "cleaned", "ratings_cleaned.csv")
```

```python
import pandas as pd

def clean_ratings(raw_path, out_path):

    # load dataset
    ratings = pd.read_csv(raw_path)
    original_rows = len(ratings)

    # remove rows without primary key
    ratings = ratings.dropna(subset='anime_id')

    # Ensure correct type for ID
    ratings['anime_id'] = pd.to_numeric(ratings['anime_id'], errors='coerce').astype(int)


    # Remove exact duplicate rows
    ratings = ratings.drop_duplicates()

    print("Rows removed:", original_rows - len(ratings))

    # Save cleaned dataset
    ratings.to_csv(out_path, index=False)

    return ratings
```
