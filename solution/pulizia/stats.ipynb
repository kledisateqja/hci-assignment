# Cleaning stats

```python
import pandas as pd
import os
```

```python
raw_path = os.path.join("../..", "dataset", "raw", "stats.csv")
out_path = os.path.join("../..", "dataset", "cleaned", "stats_cleaned.csv")
```

```python
stats = pd.read_csv(raw_path)
original_rows = len(stats)

# Drop rows without primary key
stats = stats.dropna(subset=['mal_id'])

# Convert mal_id to integer
stats['mal_id'] = pd.to_numeric(stats['mal_id'], errors='coerce').astype(int)
stats = stats.dropna(subset=['mal_id'])

# Identify numeric columns (everything except mal_id)
numeric_cols = stats.columns.drop('mal_id')

# Convert all other columns to numeric
stats[numeric_cols] = stats[numeric_cols].apply(pd.to_numeric,errors='coerce')

# Remove fully empty rows only
stats = stats.dropna(how='all')

print("Rows removed:", original_rows - len(stats))

stats.to_csv(out_path, index=False)
```
