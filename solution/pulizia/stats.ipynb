# Cleaning stats

```python
import pandas as pd
import os
```

```python
RAW_PATH = os.path.join("../..", "dataset", "raw", "stats.csv")
CLEAN_PATH = os.path.join("../..", "dataset", "cleaned", "stats_cleaned.csv")
```

```python
df = pd.read_csv(RAW_PATH)
original_rows = len(df)

# Convertiamo mal_id in numerico
df['mal_id'] = pd.to_numeric(df['mal_id'], errors='coerce')

# Senza ID non ha senso analitico
df = df.dropna(subset=['mal_id'])

# ID come intero
df['mal_id'] = df['mal_id'].astype(int)
# colonne usate
user_cols = [
    'watching', 'completed', 'on_hold',
    'dropped', 'plan_to_watch', 'total'
]

# Convertiamo tutte in numerico
df[user_cols] = df[user_cols].apply(
    pd.to_numeric, errors='coerce'
)

df = df.drop_duplicates()
df = df.drop_duplicates(subset=['mal_id'])

print("Righe iniziali:", original_rows)
print("Righe finali:", len(df))
print("Righe rimosse:", original_rows - len(df))


df.to_csv(CLEAN_PATH, index=False, encoding="utf-8")
```
