# Cleaning favs

```python
import pandas as pd
import os
```

```python
raw_path = os.path.join("../..", "dataset", "raw", "favs.csv")
out_path = os.path.join("../..", "dataset", "cleaned", "favs_cleaned.csv")
```

```python
favs = pd.read_csv(raw_path)
original_rows = len(favs)
# Keep only real columns
favs = favs[['username', 'fav_type', 'id']]
# Drop rows without valid keys
favs = favs.dropna(subset=['username', 'id'])
# Convert anime id to numeric
favs['id'] = pd.to_numeric(favs['id'], errors='coerce').astype(int)
favs = favs.dropna(subset=['id'])

# Rename for consistency
favs = favs.rename(columns={'id': 'mal_id'})
print("Rows removed:", original_rows - len(favs))

favs.to_csv(out_path, index=False)
```
