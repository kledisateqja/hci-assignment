# Cleaning favs

```python
import pandas as pd
import os
```

```python
RAW_PATH = os.path.join("../..", "dataset", "raw", "favs.csv")
CLEAN_PATH = os.path.join("../..", "dataset", "cleaned", "favs_cleaned.csv")
```

```python
df = pd.read_csv(RAW_PATH)
original_rows = len(df)

# Rendiamo espliciti i valori mancanti
df['username'] = df['username'].fillna("Unknown")
df['fav_type'] = df['fav_type'].fillna("Unknown")

# Convertiamo l'id in numerico
df['id'] = pd.to_numeric(df['id'], errors='coerce')

# Senza id non possiamo collegare l'anime
df = df.dropna(subset=['id'])

# ID come intero
df['id'] = df['id'].astype(int)
# Rimuoviamo solo righe identiche (stesso utente, stesso anime)
df = df.drop_duplicates()
print("Righe iniziali:", original_rows)
print("Righe finali:", len(df))
print("Righe rimosse:", original_rows - len(df))

df.to_csv(CLEAN_PATH, index=False, encoding="utf-8")
```
