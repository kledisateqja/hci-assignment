# Cleaning profiles

```python
import os
import pandas as pd
```

```python
#path
RAW = os.path.join(".." , ".." , "data" , "raw" , "profiles.csv")
CLEAN = os.path.join(".." , ".." , "data" , "cleaned" , "profiles_cleaned.csv")
```

```python
#load
df = pd.read_csv(RAW)
original_rows = len(df)
df.head()
```

```python
# check columns
len(df)
df.count()
```

```python
#cleaning
#username
df = df.dropna(subset=["username"])
df["username"] = df["username"].astype(str).str.strip()
df = df[df["username"] != ""]

# gender
df["gender"] = df["gender"].astype(str).str.strip()
df.loc[df["gender"].str.lower() == "nan" , "gender"] = pd.NA

# birthday
df["birthday"] = pd.to_datetime(df["birthday"], errors="coerce")

#location
df["location"] = df["location"].astype(str).str.strip()
df.loc[df["location"].str.lower() == "nan", "location"] = pd.NA

#joined
df["joined"] = pd.to_datetime(df["joined"] , errors="coerce")

#activity
activity_cols = ["watching", "completed", "on_hold", "dropped", "plan_to_watch"]
df[activity_cols] = df[activity_cols].apply(pd.to_numeric, errors="coerce")

#remove duplicates on username
before = len(df)
df = df.drop_duplicates(subset=["username"])
print("removed duplicates:", before - len(df))

#final check
len(df)
df.count()

#save
df.to_csv(CLEAN, index=False)
print("rows before:" , original_rows , "| rows after:" , len(df))
```
